```python
mcs = Configuration.max_consistent_subset
def max_reducible_subsets(cols):
    def max_reduc_sub_recursive(cols, dict, string):
        if string in dict:
            return dict[string]
        if int(string)==0:
            return []
        cols_current_set = [cols[j] for j in range(len(string)) if string[j]=='1']
        if mcs(cols_current_set)['C'][-1]>0:
            dict[string] = []
            return []
        # if we reached here, the condition mcs(cols_current_set)['C'][-1] == 0 holds
        strings_from_children = []
        for j in range(len(string)):
            if string[j]=='1':
                strings_from_child = max_reduc_sub_recursive(cols, dict, string[:j]+'0'+string[j+1:])
                for s in strings_from_child:
                    if s not in strings_from_children:
                        strings_from_children.append(s)
        if strings_from_children==[]:
            dict[string] = [string]
            return [string]
        dict[string] = strings_from_children
        return strings_from_children
    
    dict = {}
    strings = max_reduc_sub_recursive(cols, dict, '1'*len(cols))
    return (strings,[[cols[j] for j in range(len(s)) if s[j]=='1'] for s in strings])
```

```python
tt=time.time()
x,y=max_reducible_subsets(list(confs2822[0].extendable_edge_coloring_strings().keys()))
print(f"{len(x)}, {Counter([len(subset) for subset in y])}, time={time.time()-tt:.2f}")
```
